### 외부 프롬프트 로딩

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os

# 외부 YAML 파일에서 프롬프트 로드
# 코드 내부에 프롬프트를 하드코딩하지 않고 외부 파일(.yaml)에서 깔끔하게 읽어옵니다.
# 현재 노트북이 있는 폴더 경로를 가져옵니다.
# prompt_path = os.path.join(os.path.dirname(__file__), "prompts", "greeting.yaml") # py 파일의 경우

current_dir = os.getcwd() # ipynb 파일의 경우
prompt_path = os.path.join(current_dir, "prompts", "greeting.yaml")
prompt_path
```

```python
from langchain_core.prompts import load_prompt

prompt_template = load_prompt(prompt_path, encoding="utf-8")
prompt_template.template
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
```

```python
# 프롬프트 체인 구성
chain = prompt_template | llm | output_parser
```

```python
response = chain.invoke({
    "user_name": "김일남",
    "topic": "랭체인 외부 프롬프트 활용법"
})
response
```
