```python
import easyocr
from pdf2image import convert_from_path
import numpy as np
from pathlib import Path
from datetime import datetime

def pdf_to_markdown(pdf_path, output_path=None, dpi=300, languages=['ko', 'en']):
    """
    PDF 파일을 OCR로 읽어 마크다운 형식으로 변환
    
    Args:
        pdf_path: PDF 파일 경로
        output_path: 출력 마크다운 파일 경로 (None이면 자동 생성)
        dpi: 이미지 변환 DPI (기본값: 300)
        languages: OCR 언어 리스트 (기본값: ['ko', 'en'])
    
    Returns:
        str: 생성된 마크다운 파일 경로
    """
    
    # 출력 파일 경로 생성
    if output_path is None:
        pdf_name = Path(pdf_path).stem
        output_path = f"{pdf_name}_ocr.md"
    
    # EasyOCR 리더 초기화
    print(f"EasyOCR 초기화 중... (언어: {', '.join(languages)})")
    reader = easyocr.Reader(languages, gpu=False)
    
    # PDF를 이미지로 변환
    print(f"PDF를 {dpi} DPI로 변환 중...")
    images = convert_from_path(pdf_path, dpi=dpi)
    
    # 마크다운 파일 작성
    with open(output_path, 'w', encoding='utf-8') as f:
        # 헤더 작성
        f.write(f"# OCR 결과: {Path(pdf_path).name}\n\n")
        f.write(f"- **변환 일시**: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        f.write(f"- **총 페이지**: {len(images)}페이지\n")
        f.write(f"- **DPI**: {dpi}\n")
        f.write(f"- **인식 언어**: {', '.join(languages)}\n\n")
        f.write("---\n\n")
        
        # 각 페이지 처리
        for page_num, image in enumerate(images, start=1):
            print(f"페이지 {page_num}/{len(images)} 처리 중...")
            
            # 페이지 헤더
            f.write(f"## 페이지 {page_num}\n\n")
            
            # OCR 수행
            img_array = np.array(image)
            result = reader.readtext(img_array, detail=1)
            
            if not result:
                f.write("*텍스트가 감지되지 않았습니다.*\n\n")
                continue
            
            # Y 좌표 기준으로 정렬 (위에서 아래로)
            result_sorted = sorted(result, key=lambda x: x[0][0][1])
            
            # 결과를 마크다운으로 작성
            for bbox, text, confidence in result_sorted:
                # 신뢰도가 낮은 텍스트는 이탤릭으로 표시
                if confidence < 0.5:
                    f.write(f"*{text}* `(신뢰도: {confidence:.2f})`\n\n")
                else:
                    f.write(f"{text}\n\n")
            
            f.write("---\n\n")
    
    print(f"✅ 변환 완료: {output_path}")
    return output_path


# 사용 예시
if __name__ == "__main__":
    pdf_path = '../data/sample_income_tax_p30.pdf'
    output_path = pdf_to_markdown(
        pdf_path=pdf_path,
        dpi=300,
        languages=['ko', 'en']
    )
    
    print(f"\n마크다운 파일이 생성되었습니다: {output_path}")
```

```python
import easyocr
from pdf2image import convert_from_path
import numpy as np
from pathlib import Path
from datetime import datetime

def pdf_to_markdown_advanced(pdf_path, output_path=None, dpi=300, languages=['ko', 'en']):
    """
    PDF를 OCR로 읽어 구조화된 마크다운으로 변환
    """
    
    if output_path is None:
        pdf_name = Path(pdf_path).stem
        output_path = f"{pdf_name}_ocr.md"
    
    print(f"EasyOCR 초기화 중... (언어: {', '.join(languages)})")
    reader = easyocr.Reader(languages, gpu=False)
    
    print(f"PDF를 {dpi} DPI로 변환 중...")
    images = convert_from_path(pdf_path, dpi=dpi)
    
    with open(output_path, 'w', encoding='utf-8') as f:
        # 문서 헤더
        f.write(f"# {Path(pdf_path).stem}\n\n")
        f.write(f"> **변환 정보**\n")
        f.write(f"> - 변환 일시: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        f.write(f"> - 총 페이지: {len(images)}페이지\n")
        f.write(f"> - DPI: {dpi}\n")
        f.write(f"> - 인식 언어: {', '.join(languages)}\n\n")
        f.write("---\n\n")
        
        for page_num, image in enumerate(images, start=1):
            print(f"페이지 {page_num}/{len(images)} 처리 중...")
            
            f.write(f"## 📄 페이지 {page_num}\n\n")
            
            img_array = np.array(image)
            result = reader.readtext(img_array, detail=1, paragraph=False)
            
            if not result:
                f.write("> *이 페이지에서 텍스트가 감지되지 않았습니다.*\n\n")
                continue
            
            # Y 좌표로 정렬하여 라인 그룹화
            result_sorted = sorted(result, key=lambda x: x[0][0][1])
            
            current_y = None
            line_threshold = 20  # 같은 라인으로 간주할 Y 좌표 차이
            current_line = []
            
            for bbox, text, confidence in result_sorted:
                y_coord = bbox[0][1]
                
                # 새로운 라인 시작
                if current_y is None or abs(y_coord - current_y) > line_threshold:
                    # 이전 라인 출력
                    if current_line:
                        line_text = ' '.join(current_line)
                        f.write(f"{line_text}\n\n")
                    
                    current_line = [text]
                    current_y = y_coord
                else:
                    # 같은 라인에 추가
                    current_line.append(text)
            
            # 마지막 라인 출력
            if current_line:
                line_text = ' '.join(current_line)
                f.write(f"{line_text}\n\n")
            
            # 통계 정보
            total_texts = len(result)
            avg_confidence = sum(r[2] for r in result) / total_texts if total_texts > 0 else 0
            
            f.write(f"<details>\n")
            f.write(f"<summary>📊 페이지 {page_num} 통계</summary>\n\n")
            f.write(f"- 감지된 텍스트 블록: {total_texts}개\n")
            f.write(f"- 평균 신뢰도: {avg_confidence:.2%}\n")
            f.write(f"</details>\n\n")
            
            f.write("---\n\n")
    
    print(f"✅ 변환 완료: {output_path}")
    return output_path


# 사용 예시
if __name__ == "__main__":
    pdf_path = '../data/sample_income_tax_p30.pdf'
    
    # 기본 변환
    output_path = pdf_to_markdown_advanced(
        pdf_path=pdf_path,
        dpi=300,
        languages=['ko', 'en']
    )
    
    print(f"\n✨ 마크다운 파일: {output_path}")
    
    # 생성된 파일 미리보기
    print("\n📖 파일 내용 미리보기:")
    print("-" * 50)
    with open(output_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()[:20]  # 처음 20줄만 출력
        print(''.join(lines))
        if len(lines) == 20:
            print("\n... (이하 생략)")
```

```python
import easyocr
from pdf2image import convert_from_path
import numpy as np

# 간단한 원라이너 버전
reader = easyocr.Reader(['ko', 'en'], gpu=False)
images = convert_from_path('../data/sample_income_tax_p30.pdf', dpi=300)

with open('output.md', 'w', encoding='utf-8') as f:
    f.write(f"# OCR 결과\n\n")
    for i, img in enumerate(images, 1):
        f.write(f"## 페이지 {i}\n\n")
        result = reader.readtext(np.array(img))
        for _, text, conf in sorted(result, key=lambda x: x[0][0][1]):
            f.write(f"{text}\n\n")
        f.write("---\n\n")

print("✅ 완료: output.md")
```
