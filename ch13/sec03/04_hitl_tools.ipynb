```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

model = init_chat_model(
    "google_genai:gemini-3.1-flash-lite",
    temperature=0
)
```

```python
from typing import TypedDict
from typing import Annotated
from langgraph.graph.message import add_messages

class AgentState(TypedDict):
    # add_messages를 사용하여 대화 메시지 관리로 변경합니다.
    messages: Annotated[list, add_messages]
```

```python
from langchain.tools import tool

@tool
def send_email(to: str, subject: str, body: str):
    """수신자에게 이메일을 보냅니다"""

    response = interrupt({
        "action": "send_email",
        "to": to,
        "subject": subject,
        "body": body,
        "message": "이 이메일 전송을 승인하시겠습니까?",
    })

    if response.get("action") == "approve":
        final_to = response.get("to", to)
        final_subject = response.get("subject", subject)
        final_body = response.get("body", body)

        print(f"[send_email] to={final_to} subject={final_subject} body={final_body}")
        return f"Email sent to {final_to}"

    return "사용자에 의해 이메일 전송이 취소되었습니다"
```

```python
tools = [send_email]
model_with_tools = model.bind_tools(tools)
```

```python
from langgraph.graph import StateGraph, START, END
from langgraph.types import Command, interrupt


def agent_node(state: AgentState):
    result = model_with_tools.invoke(state["messages"])
    # add_messages를 사용으로 인해 result만 반환합니다.
    return {"messages": [result]}


# 조건부 엣지를 위한 함수 정의
def should_continue(state: AgentState):
    last_message = state["messages"][-1]
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "tools"
    return END
```

```python
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.prebuilt import ToolNode

tool_node = ToolNode(tools)

graph_builder = StateGraph(AgentState)

graph_builder.add_node("agent", agent_node)
# ToolNode를 Node로 추가합니다.
graph_builder.add_node("tools", tool_node)

graph_builder.add_edge(START, "agent")
# 도구 호출 여부를 위한 조건부 엣지를 추가합니다.
graph_builder.add_conditional_edges("agent", should_continue, ["tools", END])
# 도구 호출 후 다시 에이전트로 돌아가는 엣지 추가
graph_builder.add_edge("tools", "agent")

# Checkpointer로 InMemorySaver를 사용합니다.
checkpointer = InMemorySaver()
graph = graph_builder.compile(checkpointer=checkpointer)
```

```python
config = {"configurable": {"thread_id": "email-1"}}

content = (
    # 이메일 전송 요청을 경우 도구를 사용하라는 지시문 추가
    "사용자가 이메일을 보내달라고 요청하면 반드시 send_email 도구를 사용해야 합니다.\n"
    "alice@example.com에게 회의를 개최한다고 이메일을 보내주세요"
    "제목: 랭체인 앱 개발, 내용: 랭체인 앱 개발 회의를 진행하겠습니다."
)

initial = graph.invoke(
    {"messages": [{"role": "user", "content": content}]},
    config=config,
)
```

```python
state = graph.get_state(config)
```

```python
prompt = state.interrupts[0].value["message"]
prompt
```

```python
user_input = "approve" if input(prompt) == "승인" else "reject"  # 재개(resume)를 위해 사용자 입력 받기
```

```python

resumed = graph.invoke(
    ### 이메일 전송 승인 ###
    # Command(resume={"action": "approve", "subject": "업데이트된 제목"}),
    Command(resume={"action": user_input, "subject": "업데이트된 제목"}),
    ### 이메일 전송 거부 ###
    # Command(resume={"action": "reject", "subject": "업데이트된 제목"}),
    config=config,
)
# 최종 메시지의 내용을 출력
print("최종 메시지:", resumed["messages"][-1].content)
```
