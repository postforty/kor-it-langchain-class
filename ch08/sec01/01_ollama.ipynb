> LangChain Ollama: <https://docs.langchain.com/oss/python/integrations/providers/ollama>

```python
# uv add langchain-ollama

from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM
```

```python

template = """질문: {question}

답변: 차근차근 단계별로 생각해보자."""

prompt = ChatPromptTemplate.from_template(template)

model = OllamaLLM(model="gemma3:270m", base_url="http://localhost:11434") # 도커를 이용하고 있으므로 base_url을 지정해주어야 함

chain = prompt | model

answer = chain.invoke({"question": "랭체인이 무엇인가요?"})
```

```python
print(answer)
```
