```python
# FILE = "../data/sample_income_tax_p30.png"
FILE = "../data/sample_income_tax_p30.pdf"
```

```python
from docling.datamodel.base_models import InputFormat
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.pipeline.vlm_pipeline import VlmPipeline
from docling.datamodel.pipeline_options import (
    VlmPipelineOptions,
)
from docling.datamodel import vlm_model_specs

pipeline_options = VlmPipelineOptions(
    vlm_options=vlm_model_specs.SMOLDOCLING_MLX,  # <-- change the model here
)

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_cls=VlmPipeline,
            pipeline_options=pipeline_options,
        ),
    }
)

doc = converter.convert(source=FILE).document
```

```python
print(doc)
```

```python
# 텍스트 아이템 반복 출력
for text_item in doc.texts:
    print(text_item.text)

# 표(테이블) 정보 출력 예시
for table_item in doc.tables:
    print(table_item.data)  # TableData 객체 형태로 변환됨

# 그림(이미지) 정보 출력 예시
for picture_item in doc.pictures:
    print(picture_item)  # 객체 정보, 또는 picture_item.image로 이미지 접근

# 각 항목별로 상세 속성을 함께 살펴보고 싶다면
import pprint
pprint.pprint(doc.texts[0].__dict__)
```
