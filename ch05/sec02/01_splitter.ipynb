### RecursiveCharacterTextSplitter

`RecursiveCharacterTextSplitter`는 단순 분할기와 달리 **재귀적 분할 방식**을 사용해 텍스트를 나눈다.

-   단순 분할기는 정해진 규칙으로 한 번만 나누지만, 재귀적 분할기는 분할된 덩어리(chunk)가 클 경우 미리 정해둔 분할자(예: 문단, 문장, 공백) 목록을 순서대로 적용하며 덩어리를 더 작게 만든다.

-   이러한 재귀적 접근은 텍스트의 **문맥을 최대한 보존**하면서 분할하기 때문에, 덩어리가 의미를 잃지 않고 온전한 정보를 포함하게 된다.

#### chunk_size의 역할

* **`chunk_size=1000`**: 분할된 청크의 크기는 1000자를 초과할 수 없다는 **최대 길이 제한**이다. 실제 크기는 텍스트 내 구분자의 위치에 따라 결정된다.
* **분할 기준**: `["\n\n", "\n", " ", ""]`와 같은 사전에 정의된 **문자열 구분자를 계층적으로 적용**하는 기계적 프로세스이다.

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('../data/KCI_FI003153549.pdf')
pages = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000, # 최대 길이 제한
        chunk_overlap=200 # 정확히 200자를 중첩
    )
```

```python
splitted_docs = splitter.split_documents(pages)
```
