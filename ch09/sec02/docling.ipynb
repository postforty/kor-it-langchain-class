```python
FILE_PATH = "../data/sample_income_tax_p27.pdf"
```

```python
from langchain_community.document_loaders import PyPDFLoader

# 문서 로드
loader = PyPDFLoader(FILE_PATH)
documents = loader.load()
```

```python
documents # 그림으로 포함된 표는 누락됨
```

> LangChain Docling 문서: <https://python.langchain.com/docs/integrations/document_loaders/docling/>

```python
from langchain_docling import DoclingLoader

loader = DoclingLoader(file_path=FILE_PATH)
```

```python
docs = loader.load()
```

```python
for d in docs[:]:
    print(f"- {d.page_content=}") # 그림으로 포함된 표는 누락됨
```

```python
from docling.datamodel import vlm_model_specs
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import (
    VlmPipelineOptions,
)
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.pipeline.vlm_pipeline import VlmPipeline
```

```python
converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_cls=VlmPipeline,
        ),
    }
)
```

```python
doc = converter.convert(source=FILE_PATH).document # 1페이지 파싱하는데 무려 11분 이상 걸렸음
```

```python
print(doc.export_to_markdown())
```

```python
# 문서를 변환하고 결과 객체를 받음
result_doc = converter.convert(source=FILE_PATH)
```
