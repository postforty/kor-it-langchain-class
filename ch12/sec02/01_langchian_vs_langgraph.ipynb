```python
# uv add langchain

from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage

from dotenv import load_dotenv
load_dotenv()

# 모델 초기화
model = init_chat_model(
    "google_genai:gemini-2.5-flash",
    temperature=0,
)

# 메시지 구성
messages = [
    HumanMessage(content="내 이름은 김일남이야"),
]

# LLM 호출
response = model.invoke(messages)

# LLM 응답 출력
print(response.content)
```

```python
from dotenv import load_dotenv
load_dotenv()

model = init_chat_model(
    "google_genai:gemini-2.5-flash",
    temperature=0
)
```

```python
from typing import TypedDict, Annotated
from operator import add

from langgraph.graph.message import add_messages  # 매 단계마다 새로운 메시지를 추가함
from langchain_core.messages import AnyMessage # HumanMessage, AIMessage, SystemMessage, ToolMessage, FunctionMessage를 나타내는 추상화된 타입

# 상태 객체 생성용 클래스
# 딕셔너리 구조
class AgentState(TypedDict):
    messages: Annotated[AnyMessage, add_messages] # add_messages 리듀서 함수는 상태 키의 값(메시지)을 업데이트(덮어쓰는 대신 리스트에 추가)
```

```python
# 함수 노드
def generate(state: AgentState) -> AgentState:
    """
    사용자의 질문을 받아서 응답을 생성하는 노드입니다.
    """
    messages = state['messages']
    ai_message = model.invoke(messages)

    return {'messages': [ai_message]}
```

```python
from langgraph.graph import StateGraph, START, END # 상태 기반의 그래프를 생성하는 클래스(에이전트의 상태 관리, Node와 Edge 연결)

# 그래프 객체
graph_builder = StateGraph(AgentState)

# 그래스 객체에 노드 추가
graph_builder.add_node('generate', generate)

# 엣지로 노드 연결
graph_builder.add_edge(START, 'generate')
graph_builder.add_edge('generate', END)
```

```python
graph = graph_builder.compile() # 그래프를 실행 가능한 상태로 변환하는 메서드
```

```python
from IPython.display import Image, display

display(Image(graph.get_graph().draw_mermaid_png()))
```
