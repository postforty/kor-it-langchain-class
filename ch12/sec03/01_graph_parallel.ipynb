### 병렬 실행 (Parallelization)

> https://docs.langchain.com/oss/python/langgraph/workflows-agents#parallelization

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

llm = init_chat_model(
    "google_genai:gemini-3.1-flash-lite",
)
```

```python
from typing import TypedDict

# 그래프 상태 (State)
class State(TypedDict):
    topic: str
    joke: str
    story: str
    poem: str
    combined_output: str


# 노드 (Nodes)
def call_llm_1(state: State):
    """농담을 생성하는 첫 번째 LLM 호출"""

    msg = llm.invoke(f"{state['topic']}에 대한 재미있는 농담을 하나 작성해줘")
    return {"joke": msg.content}


def call_llm_2(state: State):
    """이야기를 생성하는 두 번째 LLM 호출"""

    msg = llm.invoke(f"{state['topic']}에 대한 짧은 이야기를 하나 작성해줘")
    return {"story": msg.content}


def call_llm_3(state: State):
    """시를 생성하는 세 번째 LLM 호출"""

    msg = llm.invoke(f"{state['topic']}에 대한 멋진 시를 한 편 작성해줘")
    return {"poem": msg.content}


def aggregator(state: State):
    """농담, 이야기, 시를 하나의 결과물로 종합"""

    combined = f"다음은 {state['topic']}에 대한 이야기, 농담, 그리고 시입니다!\n\n"
    combined += f"이야기(STORY):\n{state['story']}\n\n"
    combined += f"농담(JOKE):\n{state['joke']}\n\n"
    combined += f"시(POEM):\n{state['poem']}"
    return {"combined_output": combined}
```

```python
from langgraph.graph import StateGraph, START, END

# 워크플로우(그래프) 빌드
parallel_builder = StateGraph(State)

# 노드 추가
parallel_builder.add_node("call_llm_1", call_llm_1)
parallel_builder.add_node("call_llm_2", call_llm_2)
parallel_builder.add_node("call_llm_3", call_llm_3)
parallel_builder.add_node("aggregator", aggregator)

# 노드들을 연결하는 엣지 추가
parallel_builder.add_edge(START, "call_llm_1")
parallel_builder.add_edge(START, "call_llm_2")
parallel_builder.add_edge(START, "call_llm_3")
parallel_builder.add_edge("call_llm_1", "aggregator")
parallel_builder.add_edge("call_llm_2", "aggregator")
parallel_builder.add_edge("call_llm_3", "aggregator")
parallel_builder.add_edge("aggregator", END)
parallel_workflow = parallel_builder.compile()
```

```python
parallel_workflow
```

```python
# 그래프 실행 (Invoke)
state = parallel_workflow.invoke({"topic": "cats"})
print(state["combined_output"])
```
