WebBaseLoader는 기본적으로 BeautifulSoup을 사용하여 웹 페이지의 텍스트 콘텐츠를 추출합니다.

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader('https://www.langchain.com/')
docs = loader.load()
```

| 방법 | 장점 | 단점
| :-- | :-- | :--
| WebBaseLoader | 간단하게 사용 가능 | XML 태그가 그대로 포함되거나, 구조화된 데이터 추출이 어려움 |
| RSSFeedLoader | RSS 전용으로 설계됨, 각 뉴스 항목을 개별 Document로 분리 | 별도 로더 사용 필요 |

```python
# uv add feedparser newspaper3k lxml lxml_html_clean

from langchain_community.document_loaders import RSSFeedLoader
# RSS 피드 로드
loader = RSSFeedLoader(urls=["https://feeds.bbci.co.uk/news/world/rss.xml"])
docs = loader.load()
# 결과 확인
print(f"총 {len(docs)}개의 뉴스 기사 로드됨")
for doc in docs[:3]:  # 처음 3개만 출력
    print(f"제목: {doc.metadata.get('title')}")
    print(f"링크: {doc.metadata.get('link')}")
    print(f"내용: {doc.page_content[:200]}...")
    print("-" * 50)
```
