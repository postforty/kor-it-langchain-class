### RecursiveCharacterTextSplitter

재귀적 분할 (Recursive Splitting)

#### 1. 동작 원리 및 특징
| 항목 | 내용 |
| :--- | :--- |
| **핵심 개념** | 설정된 크기(`chunk_size`)보다 작아질 때까지 구분자 리스트를 순차적으로 적용하여 재귀적으로 분할 |
| **분할 전략** | 의미적 연관성이 높은 텍스트(문단 > 문장 > 단어)를 최대한 한 덩어리로 유지 |
| **기본 구분자** | `["\n\n", "\n", " ", ""]` (문단 → 줄바꿈 → 공백 → 개별 문자 순서) |
| **작동 방식** | 우선순위 높은 구분자로 나누고, 크기가 크면 다음 단계 구분자로 반복 분할 |
| **권장 용도** | 일반적인 모든 텍스트 문서 분할에 가장 추천되는 기본 방식 |

#### 2. 주요 설정 파라미터 (Parameters)
| 파라미터 | 설명 | 비고 |
| :--- | :--- | :--- |
| **`chunk_size`** | 각 청크(덩어리)의 최대 길이 | 글자 수(Characters) 기준 |
| **`chunk_overlap`** | 청크 간에 중첩되는 글자 수 | 흐름 유지를 위해 보통 10~20% 설정 |
| **`separators`** | 분할에 사용할 사용자 정의 구분자 리스트 | 다국어 처리 시 특수 문호 추가 가능 |
| **`length_function`** | 길이를 측정하는 함수 | 기본값: `len` |
| **`is_separator_regex`** | 구분자 리스트를 정규표현식으로 해석할지 여부 | 기본값: `False` |

#### 3. 주요 메서드 비교
| 메서드 | 입력 타입 | 출력 타입 | 설명 |
| :--- | :--- | :--- | :--- |
| **`split_text()`** | `str` | `List[str]` | 순수 텍스트 문자열을 분할할 때 사용 |
| **`split_documents()`** | `List[Document]` | `List[Document]` | 로더로 읽어온 문서 객체 리스트를 분할할 때 사용 |
| **`create_documents()`** | `List[str]` | `List[Document]` | 텍스트 리스트를 분할하여 메타데이터가 포함된 문서 객체로 생성 |

```python
from langchain_community.document_loaders import PDFPlumberLoader

loader = PDFPlumberLoader("../data/KCI_FI003153549.pdf")
documents = loader.load()
```

```python
# uv add langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=0)
texts = text_splitter.split_documents(documents)
```
