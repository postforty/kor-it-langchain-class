# 파이썬 PDF 라이브러리 비교: PDFPlumber, PyMuPDF, PyPDF

| 구분 | PDFPlumber | PyMuPDF (fitz) | PyPDF |
| :--- | :--- | :--- | :--- |
| **주요 특징** | **표(Table) 추출** 및 레이아웃 분석 최적화 | **압도적인 속도**와 렌더링(이미지 변환) 성능 | PDF 병합, 분할 등 **파일 조작** 및 기본 추출 |
| **작성 언어** | Pure Python (`pdfminer.six` 기반) | **C / C++** (`MuPDF` 엔진 바인딩) | Pure Python |
| **속도** | 느림 (상세 분석 과정 포함) | **매우 빠름** (C 엔진 기반) | 보통 |
| **표 추출 성능** | **최상** (객체 좌표분석 방식) | 보통 (최근 개선 중) | 낮음 (텍스트 나열 방식) |
| **이미지 변환** | 지원 (외부 도구 의존) | **자체 지원 (고해상도)** | 미지원 |
| **OCR 지원** | 미지원 (이미지 변환 후 외부 연동 필요) | **지원** (Tesseract 연동 시 내장 기능으로 수행 가능) | 미지원 |
| **라이선스** | MIT (매우 자유로움) | **GNU AGPL** (기업 사용 시 검토 필요) | BSD-3-Clause (자유로움) |
| **주 추천 용도** | 정밀한 표 데이터 추출 및 데이터화 | 대용량 문서 처리, 이미지 렌더링, OCR | 단순 텍스트 추출, PDF 관리(병합/분할) |

---

### 어떤 것을 선택해야 할까요?

*   **"PDF 안에 있는 표를 엑셀처럼 정확하게 뽑고 싶다"** 
    👉 `PDFPlumber`를 추천합니다.
*   **"수천 페이지의 PDF를 아주 빠르게 읽거나, 스캔된 문서(이미지)를 OCR로 읽어야 한다"** 
    👉 `PyMuPDF`를 추천합니다. (단, AGPL 라이선스 조건을 확인하세요.)
*   **"라이선스 걱정 없이 가볍게 PDF 파일들을 합치거나 텍스트만 간단히 뽑고 싶다"** 
    👉 `PyPDF`를 추천합니다.
